## Reporting: wragle_report


The project used three dataset in analysing @WeRateDogs twitter page:  twitter_archive_enhanced.csv, data generated by querying twitter API using tweepy named tweet-json.txt and images predicted by a neural network (predicted if the image put up for rating is a dog) named image_prediction.tsv. Twitter archive contains basic tweet information but does not contain all information needed, why twitter @WeRateDog was queried to get addition information to make up for the shortfall. The wrangling approached adopted was define, code and test in data wrangling section of the project.  The steps taken in data wrangling are discussed below. 

The datasets were gathered as follows ,  twitter_archive_enchanced.csv  was downloaded and upload,  image_prediction.tsv was already provided and twitter-json.txt was gotten querying twitter API.  All the three datasets was uploaded into jupyter notebook.

Each dataframe was assessed separately for tidiness (quality) and structural (messy) issue visually and programmatically, each issue; quality and structure was noted in a markdown cell at end of each assessment for each dataframe. 

Copies of the dataset was made, cleaned and merged into one master dataframe; twitter_archive_master.csv.

 Twitter_archive_enchanced.csv was read using pandas as df_archive and was assessed as discussed above for quality and structural issue. About eleven qualities and one structural issue were discovered. Among them was timestamp a column that recorded time and date each dog image was dropped for rating was an object datatype, this was converted to pandas datetime data type. The issue was defined as ‘wrong data format for timestamp, then coded and tested if it worked.  Missing values were dropped row wise so other column information in the index can be dropped. Irrelevant columns in df_archive were removed, first by removing the non-null values row wise as stated above and then dropping the columns. Functions were written to avoid repeating codes to facilitate getting indices and dropping columns or rows as the case may be. User rating was extracted properly using regular expression, the numerators and denominators were stored in different columns and was converted to integers after extraction. Dog states was made into one column.

Tweet-json.txt was read as tweets_df, id column in tweets_df was changed to tweets_id to facilitates merging the three dataframes.

Image_prediction.csv was read as image_pred and was cleaned for images not dog images to improve the quality. Dog breed prediction and confidence were collapsed to one column named breeds and confidence to improve the structure.


The three datasets was merged together to make one dataframe and saved as twitter_archive_master.csv. Left and inner joins were used to merge the dataframes and graphs( barchart and lineplot) were made to support insights.
